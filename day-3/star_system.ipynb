```python
import rebound
import numpy as np
import matplotlib.pyplot as plt 
```

```python
data = np.loadtxt("m.txt")
mtimes = data[:, 0]
masses = data[:, 1]

data = np.loadtxt("r.txt")
rtimes = data[:, 0]
Rsuns = data[:, 1]      # data ins Rsun units

# Convert Rsun to AU
radii = np.zeros(Rsuns.size)
for i, r in enumerate(Rsuns):
    radii[i] = r * 0.00465047

```

# Modelo de Mesa

```python
fig = plt.figure(figsize=(15,5))

ax = plt.subplot(121)
ax.set_xlabel("time ([yr]")
ax.set_ylabel("Mass [M_sun]")
# ax.set_ylabel("err_ener")
plt.plot(mtimes, masses, color="red")
# plt.plot(t1, err_ener, color="blue", label="planet 1")
ax.legend(loc="best")

ax = plt.subplot(122)
ax.set_xlabel("time [yr]")
ax.set_ylabel("Radius [au]")
# ax.set_ylabel("err_ang")
plt.plot(rtimes, radii, color="green")
# plt.plot(t1, err_ang, color="blue", label="planet 1")
ax.legend(loc="best")
```
