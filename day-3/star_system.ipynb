```python
import rebound
import numpy as np
import matplotlib.pyplot as plt 
```

```python
data = np.loadtxt("../data/m.txt")
mtimes = data[:, 0]
masses = data[:, 1]

data = np.loadtxt("../data/r.txt")
rtimes = data[:, 0]
Rsuns = data[:, 1]      # data ins Rsun units

# Convert Rsun to AU
radii = np.zeros(Rsuns.size)
for i, r in enumerate(Rsuns):
    radii[i] = r * 0.00465047
```

# Modelo de Mesa

```python
fig = plt.figure(figsize=(15,5))

ax = plt.subplot(121)
ax.set_xlabel("time ([yr]")
ax.set_ylabel("Mass [M_sun]")
plt.plot(mtimes, masses, color="red")

ax = plt.subplot(122)
ax.set_xlabel("time [yr]")
ax.set_ylabel("Radius [au]")
plt.plot(rtimes, radii, color="green")
```
